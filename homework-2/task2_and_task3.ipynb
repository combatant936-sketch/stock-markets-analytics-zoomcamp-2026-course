```python
import pandas as pd
import requests as req
```

```python
url =f"https://www.iposcoop.com/2025-pricings/"
headers = {
        'User-Agent': (
            'Mozilla/5.0 (Windows NT 10.0; Win64; x64) '
            'AppleWebKit/537.36 (KHTML, like Gecko) '
            'Chrome/58.0.3029.110 Safari/537.3'
        )
    }
from io import StringIO

def getdata(url):
    try:
            response = req.get(url, headers=headers, timeout=10)
            response.raise_for_status()

            # Wrap HTML text in StringIO to avoid deprecation warning
            # "Passing literal html to 'read_html' is deprecated and will be removed in a future version. To read from a literal string, wrap it in a 'StringIO' object."
            html_io = StringIO(response.text)
            tables = pd.read_html(html_io)

            if not tables:
                raise ValueError(f"No tables found.")

            return tables[0]

    except req.exceptions.RequestException as e:
            print(f"Request failed: {e}")
    except ValueError as ve:
            print(f"Data error: {ve}")
    except Exception as ex:
            print(f"Unexpected error: {ex}")

    return pd.DataFrame()
```

```python
df=getdata(url)
```

```python
df.columns=df.columns.str.replace(" ","_").str.lower()
```

```python
import yfinance as yf
import numpy as np
```

```python
import time
g=0
stocks_df = pd.DataFrame({'A' : []})

for i,ticker in enumerate(df_filtered["symbol"]):
  print(i,ticker)

  

  # Work with stock prices
  try:
    ticker_obj = yf.Ticker(ticker)

    # historyPrices = yf.download(tickers = ticker,
    #                    period = "max",
    #                    interval = "1d")
    df2 = ticker_obj.history(
                      period = "max",
                      interval = "1d")
    
    # sleep 1 sec between downloads - not to overload the API server
    df2["ticker"] = ticker
    df2["year"]= df2.index.year
    df2["month"] = df2.index.month
    df2["weekday"] = df2.index.weekday
    df2["date"] = df2.index.date
    time.sleep(1)
    if stocks_df.empty:
        stocks_df=df2
    else:
      stocks_df = pd.concat([stocks_df,df2], ignore_index=True)
    g=g+1
  except Exception as e:
    print(e)

print(g)
```

```python
stocks_df["growth_252d"] = (
    stocks_df["Close"] /
    stocks_df.groupby("ticker")["Close"].shift(252)
    
)
```

```python

stocks_df['volatility'] = (
    stocks_df.groupby('ticker')['Close']
    .transform(lambda x: x.rolling(30).std() * np.sqrt(252))
)
```

```python
stocks_df['Sharpe'] = (
    (stocks_df['growth_252d'] - 0.05)
    / stocks_df['volatility']
)
```

```python
final_df = stocks_df[stocks_df.date == pd.to_datetime('2026-09-11').date()]

final_df.describe()
```

```python
# 1. Feature Engineering: Calculate the 12 future growth columns
growth_cols = []
for i in range(1, 13):
    col_name = f'future_growth_{i}_m'
    growth_cols.append(col_name)
    
    # We use groupby('ticker') before shifting so we don't accidentally 
    # shift the closing price of one stock into a completely different stock!
    stocks_df[col_name] = stocks_df.groupby('ticker')['Close'].shift(-21 * i) / stocks_df['Close']
```

```python
# 2. Identify Entry Points: Find the first available trading day (min_date)
min_dates = stocks_df.groupby('ticker', as_index=False)['date'].min()
```

```python
ipo_starting_data = pd.merge(stocks_df, min_dates, on=['ticker', 'date'], how='inner')
```

```python
desc_stats = ipo_starting_data[growth_cols].describe()
```

```python
medians = desc_stats.loc['50%']
highest_growth_month = medians.idxmax()
highest_median_value = medians.max()
print(f"The highest median is in '{highest_growth_month}' with a value of {highest_median_value:.4f}")
```
